```python
import pandas as pd
import numpy as np

np.random.seed(42)

data = {
    'Order_ID': range(1, 101),
    'Category': np.random.choice(['Electronics', 'Furniture', 'Clothing'], 100),
    'Region': np.random.choice(['North', 'South', 'East', 'West'], 100),
    'Sales': np.random.randint(500, 10000, 100),
    'Profit': np.random.randint(50, 3000, 100),
    'Customer_Age': np.random.randint(18, 60, 100)
}

df = pd.DataFrame(data)

df.head()
```

```python
df.to_csv('business_data.csv', index=False)
print(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(df['Sales'], bins=20)
plt.title("Sales Distribution")
plt.show()
```

```python
sns.countplot(x='Category', data=df)
plt.title("Category Distribution")
plt.show()
```

```python
df.groupby('Category')['Sales'].sum()
df
```

```python
df.groupby('Region')['Profit'].sum()
df
```

```python
sns.heatmap(df[['Sales','Profit','Customer_Age']].corr(),
            annot=True)
plt.show()
```

```python
sns.scatterplot(x='Sales', y='Profit', data=df)
plt.show()
```
